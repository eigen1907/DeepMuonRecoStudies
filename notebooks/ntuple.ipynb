```python
from pathlib import Path

import awkward as ak
import hist
import matplotlib.pyplot as plt
import mplhep as hep
import numpy as np
import uproot

plt.style.use(hep.style.CMS)

plot_dir = Path('/users/hep/joshin/workspace/deepmuonreco/DeepMuonRecoPlot/sample')
plot_dir.mkdir(parents=True, exist_ok=True)
```

```python
def as_numpy(data):
    return ak.to_numpy(data) if isinstance(data, ak.Array) else np.asarray(data)


def plot_hist_1d(ax, data, bins, weight=None, linewidth=2.5, yerr=True,
                 label=None, histtype='step', mask=None, color=None,
                 density=None):
    data = as_numpy(data)
    if mask is None:
        mask = np.ones_like(data, dtype=bool)
    if weight is None:
        weight = np.ones_like(data)

    hist_1d = hist.Hist(
        hist.axis.Regular(*bins),
        storage=hist.storage.Weight(),
    )
    hist_1d.fill(data[mask], weight=weight[mask])
    hist_1d.plot(
        ax=ax, label=label, color=color, yerr=yerr, density=density,
        histtype=histtype, linewidth=linewidth,
    )
    ax.set_xlim(bins[1], bins[2])
    ax.set_ylim(0, None)
    return hist_1d


def save_figure(fig, sample_name, plot_name):
    output = plot_dir / f'{sample_name}_{plot_name}.png'
    fig.savefig(output, dpi=150, bbox_inches='tight')
    plt.close(fig)
    print(output)
```

```python
branches = [
    'track_qoverp',
    'track_is_trk_muon',
    'track_is_matched_muon',
    'track_match_quality',
    'tp_pdg_id',
    'tp_status',
    'dt_seg_pos_x',
    'csc_seg_pos_x',
    'gem_seg_pos_x',
    'rpc_hit_pos_x',
    'gem_hit_pos_x',
]

samples = {
    'run3': {
        'tree': tree_run3,
        'title': r'$\mu^{+}\mu^{-}$ + Run 3 PU',
        'com': 13.6,
    },
    'phase2': {
        'tree': tree_phase2,
        'title': r'$\mu^{+}\mu^{-}$ + Phase 2 PU',
        'com': 14,
    },
}

arrays = {
    name: sample['tree'].arrays(branches)
    for name, sample in samples.items()
}
```

```python
def build_counts(arr):
    dt_seg_pos_x = arr.dt_seg_pos_x
    csc_seg_pos_x = arr.csc_seg_pos_x
    gem_seg_pos_x = arr.gem_seg_pos_x
    rpc_hit_pos_x = arr.rpc_hit_pos_x
    gem_hit_pos_x = arr.gem_hit_pos_x

    n_dt_seg = ak.num(dt_seg_pos_x)
    n_csc_seg = ak.num(csc_seg_pos_x)
    n_gem_seg = ak.num(gem_seg_pos_x)
    n_rpc_hit = ak.num(rpc_hit_pos_x)
    n_gem_hit = ak.num(gem_hit_pos_x)

    return {
        'n_track': ak.num(arr.track_qoverp),
        'n_trk_muon': ak.sum(arr.track_is_trk_muon, axis=1),
        'n_matched_muon': ak.sum(arr.track_is_matched_muon, axis=1),
        'n_truth_muon': ak.sum(
            (abs(arr.tp_pdg_id) == 13) & (arr.tp_status == 1), axis=1
        ),
        'n_dt_seg': n_dt_seg,
        'n_csc_seg': n_csc_seg,
        'n_gem_seg': n_gem_seg,
        'n_rpc_hit': n_rpc_hit,
        'n_gem_hit': n_gem_hit,
        'n_seg': n_dt_seg + n_csc_seg + n_gem_seg,
        'n_hit': n_rpc_hit + n_gem_hit,
    }


counts = {name: build_counts(arr) for name, arr in arrays.items()}

for name, sample_counts in counts.items():
    print(name)
    for key, values in sample_counts.items():
        print(f'  {key}: <N> = {np.mean(as_numpy(values)):.3f}')
```

```python
def nice_upper_limit(x):
    """
    Round the upper limit to a visually convenient value.

    Examples:
        1730 -> 2000
        61   -> 100
        231  -> 250
    """
    if x <= 0:
        return 1.0

    exponent = np.floor(np.log10(x))
    scale = 10.0 ** exponent
    fraction = x / scale

    if fraction <= 1.0:
        nice_fraction = 1.0
    elif fraction <= 2.0:
        nice_fraction = 2.0
    elif fraction <= 2.5:
        nice_fraction = 2.5
    elif fraction <= 5.0:
        nice_fraction = 5.0
    else:
        nice_fraction = 10.0

    return nice_fraction * scale


def get_count_bins(
    values,
    n_bins=25,
    percentile=99.5,
    margin=1.05,
):
    """
    Determine histogram binning from a single distribution.

    The upper edge is determined from the given percentile
    to avoid a few extreme outliers dominating the x-axis.
    """
    values_np = np.asarray(as_numpy(values))

    x_max = np.percentile(values_np, percentile)
    x_max = nice_upper_limit(x_max * margin)

    return (n_bins, 0, x_max)


def get_common_count_bins(
    value_list,
    n_bins=25,
    percentile=99.5,
    margin=1.05,
):
    """
    Determine common histogram binning for multiple distributions
    plotted on the same axis.
    """
    arrays = [
        np.asarray(as_numpy(values))
        for values in value_list
    ]

    combined = np.concatenate(arrays)

    x_max = np.percentile(combined, percentile)
    x_max = nice_upper_limit(x_max * margin)

    return (n_bins, 0, x_max)


# ----------------------------------------------------------------------
# Object multiplicity
# ----------------------------------------------------------------------

object_plots = [
    (
        'n_track',
        'tracks',
        'Tracker Tracks',
        'C0',
    ),
    (
        'n_trk_muon',
        'tracker_muons',
        'Tracker Muons',
        'C1',
    ),
    (
        'n_matched_muon',
        'matched_muons',
        'Matched Muons',
        'C2',
    ),
    (
        'n_truth_muon',
        'truth_muons',
        'Truth Muons',
        'C3',
    ),
]


for sample_name, sample in samples.items():

    for count_name, plot_name, label, color in object_plots:

        values = counts[sample_name][count_name]

        # Matched/truth muons are expected to be around 0, 1, 2.
        # Keep discrete binning for these variables.
        if count_name in [
            'n_matched_muon',
            'n_truth_muon',
        ]:
            bins = (3, -0.5, 2.5)

        else:
            bins = get_count_bins(
                values,
                n_bins=25,
                percentile=99.5,
            )

        fig, ax = plt.subplots(
            figsize=(10, 8)
        )

        plot_hist_1d(
            ax=ax,
            data=values,
            bins=bins,
            color=color,
            density=True,
            label=(
                f'{label}\n'
                f'<N>: {np.mean(as_numpy(values)):.3f}'
            ),
        )

        ax.set_xlabel('Number of Objects')
        ax.set_ylabel('Normalized')

        hep.cms.label(
            'Private Work',
            ax=ax,
            data=False,
            com=sample['com'],
        )

        ax.legend(
            title=sample['title']
        )

        fig.tight_layout()

        save_figure(
            fig,
            sample_name,
            plot_name,
        )


# ----------------------------------------------------------------------
# Segment, hit, and primitive multiplicity
# ----------------------------------------------------------------------

for sample_name, sample in samples.items():

    sample_counts = counts[sample_name]

    # ------------------------------------------------------------------
    # Segments
    # ------------------------------------------------------------------

    segment_items = [
        (
            'n_dt_seg',
            'DT',
            'C0',
        ),
        (
            'n_csc_seg',
            'CSC',
            'C1',
        ),
        (
            'n_gem_seg',
            'GEM',
            'C2',
        ),
        (
            'n_seg',
            'All Segments',
            'C3',
        ),
    ]

    segment_bins = get_common_count_bins(
        [
            sample_counts[key]
            for key, _, _ in segment_items
        ],
        n_bins=25,
        percentile=99.5,
    )

    fig, ax = plt.subplots(
        figsize=(10, 8)
    )

    for key, label, color in segment_items:

        values = sample_counts[key]

        plot_hist_1d(
            ax=ax,
            data=values,
            bins=segment_bins,
            color=color,
            density=True,
            label=(
                f'{label}\n'
                f'<N>: {np.mean(as_numpy(values)):.1f}'
            ),
        )

    ax.set_xlabel('Number of Segments')
    ax.set_ylabel('Normalized')

    hep.cms.label(
        'Private Work',
        ax=ax,
        data=False,
        com=sample['com'],
    )

    ax.legend(
        title=sample['title']
    )

    fig.tight_layout()

    save_figure(
        fig,
        sample_name,
        'segments',
    )

    # ------------------------------------------------------------------
    # Hits
    # ------------------------------------------------------------------

    hit_items = [
        (
            'n_rpc_hit',
            'RPC',
            'C0',
        ),
        (
            'n_gem_hit',
            'GE1/1 & GE2/1',
            'C1',
        ),
        (
            'n_hit',
            'All Hits',
            'C2',
        ),
    ]

    hit_bins = get_common_count_bins(
        [
            sample_counts[key]
            for key, _, _ in hit_items
        ],
        n_bins=25,
        percentile=99.5,
    )

    fig, ax = plt.subplots(
        figsize=(10, 8)
    )

    for key, label, color in hit_items:

        values = sample_counts[key]

        plot_hist_1d(
            ax=ax,
            data=values,
            bins=hit_bins,
            color=color,
            density=True,
            label=(
                f'{label}\n'
                f'<N>: {np.mean(as_numpy(values)):.1f}'
            ),
        )

    ax.set_xlabel('Number of Hits')
    ax.set_ylabel('Normalized')

    hep.cms.label(
        'Private Work',
        ax=ax,
        data=False,
        com=sample['com'],
    )

    ax.legend(
        title=sample['title']
    )

    fig.tight_layout()

    save_figure(
        fig,
        sample_name,
        'hits',
    )

    # ------------------------------------------------------------------
    # All primitives
    # ------------------------------------------------------------------

    primitive_counts = {
        'Segments': sample_counts['n_seg'],
        'Hits': sample_counts['n_hit'],
        'All Primitives': (
            sample_counts['n_seg']
            + sample_counts['n_hit']
        ),
    }

    primitive_bins = get_common_count_bins(
        list(primitive_counts.values()),
        n_bins=25,
        percentile=99.5,
    )

    fig, ax = plt.subplots(
        figsize=(10, 8)
    )

    for (
        (label, values),
        color,
    ) in zip(
        primitive_counts.items(),
        ['C0', 'C1', 'C2'],
    ):

        plot_hist_1d(
            ax=ax,
            data=values,
            bins=primitive_bins,
            color=color,
            density=True,
            label=(
                f'{label}\n'
                f'<N>: {np.mean(as_numpy(values)):.1f}'
            ),
        )

    ax.set_xlabel('Number of Primitives')
    ax.set_ylabel('Normalized')

    hep.cms.label(
        'Private Work',
        ax=ax,
        data=False,
        com=sample['com'],
    )

    ax.legend(
        title=sample['title']
    )

    fig.tight_layout()

    save_figure(
        fig,
        sample_name,
        'primitives',
    )
```
